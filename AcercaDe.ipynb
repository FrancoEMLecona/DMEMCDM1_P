### Llaves Principales
>- homicide_rate, firearm_homicide_rate (por 100k), firearm_homicides
>- firearm_share_of_homicides_pct
>- firearms_per_100_people
>- Contexto desde el Banco Mundial: poblacion, gdp por cabeza, index Gini, desempleos, urbanizacion, porcentaje de hombres entre 15-64 años, gasto militar, enlistamiento a la secundaria (tasa bruta)


##### Dato curioso

> Alrededor de **95 países**, la correlación de **Spearman** entre las armas por cada 100 habitantes y la tasa de homicidios cometidos con armas de fuego es cercana a **0**, lo que sugiere una ausencia de relación monotónica significativa entre ambas variables.

#### Acerca del género y las parejas

> En esta versión se agregaron las tasas de **feminicidio** y **homicidio**, junto con las tasas de **asesinatos cometidos por la pareja íntima** (UNODC) para aproximadamente **90 países**.

> A nivel mundial, el promedio anual de homicidios es aproximadamente **3.2 veces mayor** que el de feminicidios. Sin embargo, los feminicidios cometidos por la pareja son cerca de **2.9 veces más frecuentes** que los homicidios de hombres cometidos por sus parejas.

#### Observaciones
- Homicide = homicidio intencional (UNODC ICCS definition); suicidios y accidentes no estan incluidos.
- Valores perdidos significan que el recurso original no se encuentra registrado, **no es cero**
- El reporte de Weapon (Mechanism) esta incompleto para algunas ciudades, por ejemplo, el UNODC de Brazil solo cuenta con una parte de sus homicidios, por lo que **firerm_share_of_homicides_pct** es incompleto comparandolo con **homicide_rate**
- El conteo de armas son estimados con valores inconciertos, estos se basan en **computation_method**
